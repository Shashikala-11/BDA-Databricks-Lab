# Lab 04: DataFrame Transformations in Databricks

#### Load Sample Dataset

```python
df=spark.read.csv('/Workspace/Users/shashikalagupta.rajesh@gmail.com/Drafts/Expanded_data_with_more_features.csv',header=True,inferSchema=True)

df.show(5)
```

#### Filter Rows

```python
# Filter students with math score >80
high_math=df.filter(df.MathScore>80)
high_math.show(5)
```

#### Select Specific Columns

```python
# select id and reading score
df.select("_c0","ReadingScore").show(5)
```

#### Average scores by gender

```python
avg_scores=df.groupBy("Gender").agg(
    {"MathScore":"avg","ReadingScore":"avg","WritingScore":"avg"}
)
display(avg_scores)
```

#### Join Two DataFrames

```python
# Create another Dataframe with student hobbies
hobbies=[
    (1,"Football"),
    (2,"Chess"),
    (3,"Music"),
    (4,"Dance")
]

hobby_df=spark.createDataFrame(hobbies,['_c0','hobby'])
```

```python
# Join with main DataFrame
joined_df=df.join(hobby_df,"_c0","left")
joined_df.show()
```

```python
# Top 5 students writing score
df.orderBy(df.WritingScore.desc()).show(5)
```
